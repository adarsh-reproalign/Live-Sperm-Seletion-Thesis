```python
from ultralytics import YOLO

model = YOLO("yolo26m.pt")

model.train(
    data="Sperm-Detection-VISEM-1/data.yaml",   
    epochs=100,
    imgsz=640,
    batch=8,
    box=7.5,
    cls=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    mosaic=1.0,
    device=None,   
    amp=True,   
    project="runs/train",
    name="v26_640_m"
)

print("Training completed!")
```
