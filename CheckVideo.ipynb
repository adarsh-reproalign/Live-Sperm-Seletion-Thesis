```python
import cv2
import pandas as pd
from inference_sdk import InferenceHTTPClient

CLIENT = InferenceHTTPClient(
    api_url="https://serverless.roboflow.com",
    api_key="ANLwTWOL88uZwQ3Pm3tO"
)
MODEL_ID = "sperm-detection-visem/1"

video_path = "Sperm_video.mp4"

cap = cv2.VideoCapture(video_path)

data = []
frame_id = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Save temporary frame
    temp_frame = "temp.jpg"
    cv2.imwrite(temp_frame, frame)

    result = CLIENT.infer(temp_frame, model_id=MODEL_ID)

    predictions = result["predictions"]

    for obj_id, pred in enumerate(predictions):

        data.append([
            frame_id,
            obj_id,
            pred["x"],
            pred["y"],
            pred["width"],
            pred["height"]
        ])

    print("Processed frame:", frame_id)

    frame_id += 1

cap.release()

df = pd.DataFrame(
    data,
    columns=[
        "frameid",
        "trackerid",
        "x_center",
        "y_center",
        "width",
        "height"
    ]
)

df.to_csv("ground_truth.csv", index=False)

print("Ground truth generated successfully")
```

```python
import cv2
import os

image_folder = "frame"
video_name = "Sperm_video.mp4"

images = sorted(os.listdir(image_folder))

# read first image to get size
frame = cv2.imread(os.path.join(image_folder, images[0]))
height, width, layers = frame.shape

video = cv2.VideoWriter(
    video_name,
    cv2.VideoWriter_fourcc(*'mp4v'),
    10,  
    (width, height)
)

for image in images:
    img_path = os.path.join(image_folder, image)
    frame = cv2.imread(img_path)
    video.write(frame)

video.release()

print("Video created successfully")
```
